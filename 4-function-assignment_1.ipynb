```python
# Create a function that checks whether the given number is Odd or Even
def OddEven():
    num=int(input("Enter a number: "))
    print("Enter a number: ",num)
    if(num % 2 == 0):
        return f"{num} is Even number"
    else:
        return f"{num} is Odd number"

Odd_Even=OddEven()
print(Odd_Even)
```

```python
# Create a function that tells eligibility of marriage for male and female according to their age limit like 21 for male and 18 for female
def Eligible():
    gender = input("Your Gender:")
    print(f"Your Gender:{gender}")
    age = int(input("Your Age:"))
    print(f"Your Age:{age}")
    if (gender.lower()=="male" and age >= 21):
        return f"ELIGIBLE"
    elif (gender.lower()=="female" and age >= 18):
        return f"ELIGIBLE"
    else:
        return f"NOT ELIGIBLE"

EligiblityForMarriage=Eligible()
print(EligiblityForMarriage)
```

```python
# calculate the percentage of your 10th mark
def percentage():
    sub1=int(input("Subject1= "))
    print("Subject1= ",sub1)
    sub2=int(input("Subject2= "))
    print("Subject2= ",sub2)
    sub3=int(input("Subject3= "))
    print("Subject3= ",sub3)
    sub4=int(input("Subject4= "))
    print("Subject4= ",sub4)
    sub5=int(input("Subject5= "))
    print("Subject5= ",sub5)
    total=sub1+sub2+sub3+sub4+sub5
    print("Total : ",total)
    percent=total/5
    return f"Percentage: {percent:3f}"

Findpercent = percentage()
print(Findpercent)
```

```python
#print area and perimeter of triangle using functions
def triangle():
    print(f"Area of Triangle Inputs")
    print("-"*10)
    height =int(input("Height:"))
    print("Height:",height)
    base=int(input("Base:"))
    print("Base:",base)
    area_formula= (height * base)/2
    
    print(f"Perimeter of Triangle Inputs")
    print("-"*10)
    height1 =int(input("Height1:"))
    print("Height1:",height1)
    height2 =int(input("Height2:"))
    print("Height2:",height2)
    breadth =int(input("breadth:"))
    print("breadth:",breadth)
    perimeter_formula =height1+ height2+ breadth
    return area_formula, perimeter_formula

area_formula,perimeter_formula =triangle()
print(f"Area of Triangle:{area_formula}")
print(f"Perimeter of Triangle: {perimeter_formula}")
```
