```python
### BMI Index calculation
bmi_index=float(input("Enter the BMI Index: "))
bmi_index= round(bmi_index,2)
print("Enter the BMI Index",bmi_index)
if(bmi_index < 18.5):
    print("Underweight")
elif(bmi_index >= 18.5 and bmi_index <= 24.9):
    print("Healthy Weight")
elif(bmi_index >= 25.0 and bmi_index <= 29.9):
    print("Overweight")
elif(bmi_index >= 30.0 and bmi_index <= 34.9):
    print("Very Overweight or Obesity (Class 1) ")
elif(bmi_index >= 35.0 and bmi_index <= 39.9):
    print("Obesity (Class 2)")
else:
    print("Obesity (Class 3)")
```
